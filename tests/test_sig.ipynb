```python
import pysubgroup as ps
from pysubgroup.datasets import get_credit_data

data = get_credit_data()
data = data.map(lambda x: x.decode('utf-8') if isinstance(x, bytes) else x)

task = ps.SubgroupDiscoveryTask(
    data,
    ps.BinaryTarget('class', 'good'),
    ps.create_selectors(data, ignore=["class"]),
    result_set_size=10,
    depth=2,
    qf=ps.WRAccQF()
)

wrapper = ps.SignificanceDecorator(ps.BeamSearch(), num_permutations=200)
wrapped_result = wrapper.execute(task)
wrapped_result.to_dataframe()
```

```python
import pysubgroup as ps 
from pysubgroup.datasets import get_credit_data

data = get_credit_data()
data = data.map(lambda x: x.decode('utf-8') if isinstance(x, bytes) else x)

task = ps.SubgroupDiscoveryTask(
    data, 
    ps.NumericTarget("credit_amount"), 
    ps.create_selectors(data, ignore=["credit_amount"]), 
    result_set_size=10, 
    depth=3, 
    qf=ps.StandardQFNumeric(1.0)
)

wrapper = ps.SignificanceDecorator(ps.BeamSearch(), num_permutations=200)
result = wrapper.execute(task)
result.to_dataframe()
```

```python
import pickle
import pysubgroup as ps
from pysubgroup.datasets import get_credit_data
data = get_credit_data()
data = data.map(lambda x: x.decode('utf-8') if isinstance(x, bytes) else x)

target = ps.BinaryTarget('class', 'good')
searchspace = ps.create_selectors(data, ignore=["class"])
task = ps.SubgroupDiscoveryTask(
    data, target, searchspace, result_set_size=10, depth=2, qf=ps.WRAccQF()
)

def test_pickling(task):
    try:
        pickle.dumps(task)  # Attempt to serialize
        print("Task is picklable!")
    except Exception as e:
        print(f"Pickle failed: {str(e)}")

test_pickling(task)
```
